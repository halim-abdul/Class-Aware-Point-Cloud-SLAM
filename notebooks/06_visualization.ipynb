# Inspect trajectory and semantic landmarks
Synthetic experiment; no external dataset required.

```python
import matplotlib.pyplot as plt
import numpy as np
from class_slam.demo import sequence
from class_slam import SLAM
slam=SLAM()
for scan,_ in sequence():
    slam.add(scan)
slam.optimize()
cloud=slam.map(); poses=np.asarray(slam.poses)
fig,ax=plt.subplots(figsize=(7,5))
ax.scatter(cloud.points[:,0],cloud.points[:,1],c=cloud.labels,s=2,alpha=0.3)
ax.plot(poses[:,0,3],poses[:,1,3],color='red',label='Estimated trajectory')
ax.set(xlabel='x [m]',ylabel='y [m]',aspect='equal')
ax.legend(); plt.show()
```
