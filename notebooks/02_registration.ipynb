# Recover a known 3D transform
Synthetic experiment; no external dataset required.

```python
import numpy as np
from class_slam.demo import sequence
from class_slam import icp
a, _ = next(sequence())
a = a.static()
T = np.eye(4); T[:3,3] = [0.1, -0.03, 0.02]
r = icp(a, a.transform(T))
print(r)
np.testing.assert_allclose(r.transform, T, atol=1e-5)
```
